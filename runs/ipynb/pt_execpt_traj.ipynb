```python
from pathlib import Path
import torch, shutil

src = Path("/dataset/gmdit/train1.4_1k_traj")   # 원본 .pt 폴더
dst = Path("/dataset/gmdit/train1.4_1k")   # 저장할 폴더
dst.mkdir(parents=True, exist_ok=True)

for f in src.glob("*.pt"):   # 필요하면 "**/*.pt" 로 재귀 탐색
    obj = torch.load(f, map_location="cpu")
    if isinstance(obj, dict) and "traj" in obj:
        obj.pop("traj", None)
    out_path = dst / f.name
    torch.save(obj, out_path)

print("done ✅")
```
