```python
from pathlib import Path
import random
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

def show_random_grid(d, seed=None):
    """d: 이미지들이 들어있는 디렉토리 경로"""
    if seed is not None:
        random.seed(seed)
    exts = {'.png', '.jpg', '.jpeg', '.webp', '.bmp', '.gif'}
    files = [p for p in Path(d).rglob('*') if p.suffix.lower() in exts]
    if not files:
        raise ValueError(f'No images found under: {d}')

    k = 25
    picks = random.sample(files, k) if len(files) >= k else random.choices(files, k=k)

    fig, axes = plt.subplots(5, 5, figsize=(10, 10))
    for ax, p in zip(axes.ravel(), picks):
        try:
            ax.imshow(mpimg.imread(p))
        except Exception:
            ax.text(0.5, 0.5, 'load fail', ha='center', va='center')
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
show_random_grid('/dataset/val2014')
```
