```python
import torch
weights = torch.load('samplings/DiT/1.5/200/Euler/100/paper_0/13.pt')
weights.keys()
```

```python
t = weights['timesteps'][:-1]
alphas = weights['alphas'][:-1]
sigmas = weights['sigmas'][:-1]
preds = weights['preds'][:, 1, 0, 16, 16]
print(t.shape)

import matplotlib.pyplot as plt

plt.scatter(t, preds)
plt.gca().invert_xaxis()  # x축 방향 반전
plt.show()

plt.scatter(torch.log(alphas/sigmas), preds)
plt.show()

plt.scatter(sigmas/alphas, preds)
plt.gca().invert_xaxis()  # x축 방향 반전
plt.show()
```
