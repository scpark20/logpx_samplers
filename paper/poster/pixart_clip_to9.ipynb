```python
import matplotlib.pyplot as plt
import numpy as np

steps = np.array([3, 4, 5, 6, 7, 8, 9])

# RN101 CLIP values from your table
results = {
    "Dual-Solver": np.array([0.4499, 0.4732, 0.4784, 0.4803, 0.4806, 0.4814, 0.4815]),
    "BNS-Solver":      np.array([0.4320, 0.4582, 0.4694, 0.4733, 0.4746, 0.4755, 0.4757]),
    "DS-Solver":       np.array([0.4303, 0.4568, 0.4692, 0.4748, 0.4754, 0.4762, 0.4764]),
    "DPM-Solver++":    np.array([0.4422, 0.4676, 0.4746, 0.4763, 0.4768, 0.4771, 0.4771]),
    "DDIM":            np.array([0.4469, 0.4670, 0.4739, 0.4764, 0.4763, 0.4778, 0.4779]),
}

style = {
    "Dual-Solver": dict(color="#FF2D55", lw=2.2, ls="-",  marker="o", zorder=3),
    "DS-Solver":       dict(color="#00B8FF", lw=2.2, ls="-",  marker="s", zorder=3),
    "BNS-Solver":      dict(color="#00E676", lw=1.6, ls="-",  marker="^", zorder=2),
    "DPM-Solver++":    dict(color="#000000", lw=1.2, ls="--", marker="P", zorder=1),  # black
    "DDIM":            dict(color="#9AA0A6", lw=1.2, ls="--", marker="X", zorder=1),  # gray
}

fig, ax = plt.subplots(figsize=(5, 4.0), dpi=600)
lines = {}

for label, vals in results.items():
    st = style[label]
    (line,) = ax.plot(
        steps, vals,
        label=label,
        marker=st["marker"],
        linewidth=st["lw"],
        linestyle=st["ls"],
        color=st["color"],
        markersize=4,
        zorder=st["zorder"],
        alpha=0.9,
    )
    lines[label] = line

ax.set_xlabel("Steps")
ax.set_ylabel("CLIP ↑")
ax.set_xticks(steps)
ax.set_xlim(2.5, 9.5)
ax.grid(True, which="both", linestyle=":", linewidth=0.8, alpha=0.7)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

leg = ax.legend(handles=[lines[k] for k in results.keys()],
                frameon=False, loc="lower right")
for txt in leg.get_texts():
    if txt.get_text().startswith("Dual-Solver"):
        txt.set_fontweight("bold")

fig.tight_layout()
plt.show()
```
