```python
import numpy as np
import torch
from scipy import linalg

original = np.load('/data/scpark/fid/VIRTUAL_imagenet256_labeled.npz')
stats = torch.load('/data/scpark/fid/VIRTUAL_imagenet256_labeled_stats.pt')

def calc_fid(mu1, sigma1, mu2, sigma2, eps=1e-6):
    mu1 = np.atleast_1d(mu1).astype(np.float64)
    mu2 = np.atleast_1d(mu2).astype(np.float64)
    sigma1 = np.atleast_2d(sigma1).astype(np.float64)
    sigma2 = np.atleast_2d(sigma2).astype(np.float64)

    diff = mu1 - mu2
    covmean, _ = linalg.sqrtm(sigma1 @ sigma2, disp=False)
    if not np.isfinite(covmean).all():
        I = np.eye(sigma1.shape[0], dtype=np.float64) * eps
        covmean = linalg.sqrtm((sigma1 + I) @ (sigma2 + I))[0]
    if np.iscomplexobj(covmean):
        covmean = covmean.real
    return float(diff @ diff + np.trace(sigma1 + sigma2 - 2 * covmean))

calc_fid(original['mu'], original['sigma'], stats['mu'], stats['sigma'])
```
