```python
from scipy import linalg
import numpy as np
import os, torch
from tqdm import tqdm
from reports.util import load_config
from pathlib import Path

def get_clip_score(d, model='ViT-L/14'):
    s = n = 0
    for p in tqdm(Path(d).rglob('*.pt')):
        try:
            s += torch.load(p, map_location='cpu')['clip_score_' + model].float().mean().item()
            n += 1
        except Exception:
            continue
    if n == 0:
        raise ValueError(f'No clip_score found under: {d}')
    return s / n, n  # (average, file_count)
```

```python
pt_dirs = [
           #'samplings/SANA/4.5/6/Euler/30000/euler_clip/euler_clip_0',
           #'samplings/SANA/4.5/5/Euler/30000/euler_clip/euler_clip_0',
           #'samplings/SANA/4.5/4/Euler/30000/euler_clip/euler_clip_0',
           'samplings/SANA/4.5/3/Euler/30000/euler_clip/euler_clip_0',
        ]

for pt_dir in pt_dirs:
    if not os.path.exists(pt_dir):
        continue
    score1, num = get_clip_score(pt_dir, model='ViT-L/14')
    score2, num = get_clip_score(pt_dir, model='ViT-L/14@336px')
    print(pt_dir, score1, score2, num)
```
