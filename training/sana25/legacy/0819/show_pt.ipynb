```python
import torch
weights = torch.load('samplings/sana/valid4.5/sana_valid4.5_10/0.pt')

import matplotlib.pyplot as plt
import torch

def show(weights):
    x = weights['sample']            # (C,H,W) = (32,32,32)
    # per-channel min-max normalize to [0,1]
    x = (x - x.amin(dim=(1,2), keepdim=True)) / (x.amax(dim=(1,2), keepdim=True) - x.amin(dim=(1,2), keepdim=True) + 1e-8)

    fig, axs = plt.subplots(4, 8, figsize=(12, 6))
    for i, ax in enumerate(axs.flat):
        ax.imshow(x[i].cpu(), cmap='gray', vmin=0, vmax=1)
        ax.set_title(f'{i}', fontsize=8)
        ax.axis('off')
    plt.tight_layout()
    plt.show()
```

```python
weights1 = torch.load('samplings/sana/valid4.5/sana_valid4.5_10/0.pt')
weights2 = torch.load('/dataset/sana/valid4.5_5090/0.pt')
show(weights1)
show(weights2)
```
