```python
import json, random
import numpy as np

def load_and_shuffle_coco_captions(path, seed=42, size=50000):
    with open(path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    prompts = [(int(ann['image_id']), ann['caption']) for ann in data['annotations']]
    random.seed(seed)
    random.shuffle(prompts)
    return prompts[:size]

```

```python
subset = load_and_shuffle_coco_captions('annotations/captions_train2014.json', size=50000)
np.savez_compressed('mscoco2014_train.npz', subset)
subset = load_and_shuffle_coco_captions('annotations/captions_val2014.json', size=10000)
np.savez_compressed('mscoco2014_valid.npz', subset)
```
